# Question 04: [ Marks 15 ]

Create a scalar tensor x = 3.0 with gradient tracking enabled. <br>
Compute y = x² + 4x. <br>
Perform backpropagation. <br>
Print the gradient of x. <br>

Write the full PyTorch code yourself. <br>

## Answer 04:

```python
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x**2 + 4*x

y.backward()

print(x.grad)
```

# Question 05: [ Marks 10 ]

Create two tensors:
w = 2.0 (requires_grad=True) <br>
b = 1.0 (requires_grad=True) <br>

Let input x = 3.0. <br>
Compute z = w*x + b. <br>
Use Sigmoid to get y_pred.
 <br>

Define Binary Cross Entropy manually (without built-in loss). <br>
Compute gradients using backward(). <br>
Print gradients of w and b. <br>

Write full code.

## Answer 05:

```python
import torch

w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

x = torch.tensor(3.0)
y = torch.tensor(1.0) # assume lable y=1
z = w*x + b

y_pred = torch.sigmoid(z)

L = -( y * torch.log(y_pred) + (1-y) * torch.log(1-y_pred) )

L.backward()

print(w.grad)
print(b.grad)
```

# Question 07: [ Marks 20 ]


Create tensor w = 2.0 (requires_grad=True). <br>

Compute two different losses sequentially and call backward() twice. <br>
Print w.grad. <br>

Do not clear gradients. <br>
Explain what you observe. <br>

Write full code.


## Answer 07:



```python
import torch

w = torch.tensor(2.0, requires_grad=True)

#assume
loss1 = w**2
loss2 = w**3

loss1.backward()
print(w.grad)

loss2.backward()
print(w.grad)
```

After the first loss backward call the gradient of $w$  becomes 4 since $\frac{d}{dw}(w^2) = 2w = 4$.
but when the second loss backward is called then instead of printing 12 as second gradient, the code prints 16 , it adds the first gradient's value with the new second calculation.

$\frac{d}{dw}(w^3) = 3w^2 = 12$ but we get  $4_{previous} +  12_{current}= 16$
