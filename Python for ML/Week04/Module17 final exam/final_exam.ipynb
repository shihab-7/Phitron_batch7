Answer to Q1

```python
import numpy as np
import pandas as pd

df = pd.read_csv('final-employee-ds.csv')

numpy_array = np.array(df["TrainingHours"])
numpy_array = np.astype(numpy_array, float)
print(np.mean(numpy_array))
print(np.std(numpy_array))
```

```python
# df.info()
# print(df.groupby("Department")["Salary"].median())
df["Salary"] = df["Salary"].fillna(df.groupby("Department")["Salary"].transform("median"))
Ex_12_P_5 = df[((df["ExperienceYears"]>12)&(df["ProjectCount"]>5))]
total_salary = Ex_12_P_5["Salary"].sum()
print(total_salary)
```

Answer to Q3

```python
newdf = df[(((df["Department"]=="IT")|(df["Department"]=="Finance")) & ((df["Age"]>=30)&(df["Age"]<=45)) & (df["PerformanceScore"]>88))]
newdf.sort_values("Salary", ascending=False)
print(newdf[["Name", "Department", "Salary", "PerformanceScore"]])
```

Answer to Q4

```python
df["SalaryPerHour"] = df["Salary"] / (df["WorkHoursPerWeek"]*4)
df.sort_values("SalaryPerHour", ascending=False, inplace=True)
tmp = df.head(5)
print(tmp[["Name", "SalaryPerHour"]])
```
