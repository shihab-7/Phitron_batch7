loop in python

```python
for i in range(10):
    print(i)

# range(start, end, step)
# range(10) returns a list starting from 0 to 9
print("\n")
for i in range(1, 11, 3):
    print(i)

# nested for
for i in range(5):
    for j in range(5):
        print(i, j)
    print("\n")
```

```python
# sum of digits using while loop
num=210
tmp=0
while num>0:
    tmp+=num%10
    num//=10
print(tmp)
```

continue and break

```python
sum=0
for i in range(1, 11):
    print(i, "currently")

    if i%2==0:
        continue
    sum+=i
    print(i, "added to sum", sum)

# continue is used to skip
# break is used to stop the loop

sum1=0
for i in range(1, 11):
    print(i, "currently")

    sum1+=i
    if sum1 > 7:
        break
    print("sum is not more than 7:", sum1)
```

min max cf judge

```python
nums=input()
nums=nums.split()
a=int(nums[0])
b=int(nums[1])
c=int(nums[2])

min = a; max = a

if max < b:
    max=b
if max < c:
    max=c
if min > b:
    min=b
if min > c:
    min=c

print(min, max)
```

even, odd, positive, negative cf problem

```python
inp = int(input())
nums=input().split()
even=0; odd=0; positive=0; negative=0

for i in range(inp):
    num=int(nums[i])
    if num%2==0:
        even+=1
    else:
        odd+=1
    if num>0:
        positive+=1
    elif num<0:
        negative+=1
print(even, odd, positive, negative)
```
