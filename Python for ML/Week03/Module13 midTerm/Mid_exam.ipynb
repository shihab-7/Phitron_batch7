```python
# You must run this code and use that data set for all the questions
import numpy as np
import pandas as pd

np.random.seed(42)

ids = np.arange(1, 11)
ages = np.random.randint(18, 60, 10)
salaries = np.random.randint(30000, 90000, 10)
departments = np.array(["HR", "IT", "Finance", "IT", "HR", "Sales", "Finance", "IT", "Sales", "HR"])

DF = pd.DataFrame({
"id": ids,
"age": ages,
"salary": salaries,
"dept": departments
})

DF.to_csv("employees.csv", index=False)
print("Sample Data Created and Saved as employees.csv")
```

Answer to Q1

```python
Numpy_array_ages=np.array(ages)
Numpy_array_salaries=np.array(salaries)

print("Ages:", Numpy_array_ages)
print(Numpy_array_ages.dtype)
print(Numpy_array_ages.ndim)
print(Numpy_array_ages.shape)

print("Salaries:", Numpy_array_salaries)
print(Numpy_array_salaries.dtype)
print(Numpy_array_salaries.ndim)
print(Numpy_array_salaries.shape)
```

Answer to Q2

```python
salaries_array=np.array(salaries, dtype=np.int64)
highest_salary=np.max(salaries_array)
lowest_salary=np.min(salaries_array)
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)

average_salary=np.mean(salaries_array)
average_age=np.mean(ages)
print("Average Salary:", average_salary)
print("Average Age:", average_age)
```
