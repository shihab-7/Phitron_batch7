# Question 01: [ Marks 10 ]
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

## Answer 01:

Gradient descent is an optimization algorithm. It is used for minimizing loss. In gradient descent the weights and biases are adjusted step by step in the direction that reduces the loss function. The algorithm works by calculating the slope of the loss function with respect to each parameter. The slope tells us how much the loss changes if a parameter slightly changes.
The update rule eqation is :
$$ w_{\text{new}} = w_{\text{old}} - \eta \frac{\partial L}{\partial w} $$
Here,
- $w$ = weight
- $\eta$ = learning rate
- $L$ = loss function
- $\frac{\partial L}{\partial w}$ = gradient of loss with respect to weight

At first the whole process starts with random weights. Predictions are calculated using forward propagation and loss is also calculated. Then using backpropagation the gradient is calculated. Using the gradient descent formula the weights are updated. This whole process is repeated until the loss become small.

The learning rate controls the step size during weight updates. If the learning rate is too high then the update become very large, the model may cross the minimum loss point. If the learning rate is too low then the training become slow and the model can get stuck before reaching the optimal solution.



# Question 02: [ Marks 10 ]

What is forward propagation in an artificial neural network? Explain step by step how data moves from the input layer to the output layer.

## Answer 02:

In an artificial neiral network the forward propagation is the process where input data passes through the neural network layer by layer.
- The input layer only passes data forward. As example from a dataset the input layer receives the feature like age, height, earnings etc.
- After receiving the inputs the weighted sum is calculated. Each neuron multiplies inputs by weights and adds a bias.
- The weighted sum is passed through an activation function like sigmoid or ReLU to provide non-linearity into the network.
- In the hidden layers output from one layer becomes the input of the next layer. These hidden layers helps the model to learn complex patterns.
- Then finally the output layer provides the prediction.

# Question 04: [ Marks 15 ]

Explain the sigmoid activation function. Where is it used? What are its advantages and disadvantages? What problem does it cause during training?

## Answer 04:

The sigmoid activation function is :
$$\sigma(x) = \frac{1}{1+e^{-x}}$$
This function converts any input value into a number between 0 and 1. It is used for binary classification problems.

Advantages :      
- It works perfectly for simple binary classifications
- It is simple and differentiable
- Output is easy to interpret

Disadvantages :     
- Outputs are not zero centerd
- Computationally expensive for exponetial calculation
- Saturates for very large and small values

During training the sigmoid function causes vanishing gradient problem. For very large positive or negative inputs the gradient becomes very small, the weight updates also become extremly small. The training become very slow, as a result for deep networks using sigmoid is not effective.

# Question 05: [ Marks 10 ]
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Answer 05:


Softmax is an activation function used in multi-class classification problems. It converts raw outputs into probabilities whose total sum equals 1.

Softmax is used in multi-class classification because it compares multiple classes together, produce normalized probabilities and ensure total probability equals 1.
Softmax is totally different from sigmoid because sigmoid is used for binary classification when softmmax classifies multiple classes . Sigmoid outputs independent probabilities but softmax ouputs related probabilities. Sigmoid output is in a range between 0 to 1 and softmax outputs sum to 1. For sigmoid one output neuron is enough but for softmax multiple output neuron is required .
As an example , if we want yes or no result we need to use sigmoid , if we need to classify between animals like cat, dog, cow we need to use softmax.

# Question 06: [ Marks 20 ]

Explain the structure of an Artificial Neural Network in detail. Describe the role of the input layer, hidden layers, and output layer. How do weights, biases, and activation functions work together to transform input data into predictions? Illustrate your explanation with a simple example.

## Answer 06:

An Artificial Neural Network is a computational model consist with multiple neuron as similar as a human brain. Below the structure and the details of each components are given :
- **Input layer** : It receives the data coming from the real world, for example numeric values from a weather dataset like temperature, rain, humidity etc.
- **Hidden layers** : Here multiple neurons together forms a layer where each neuron extracts different features from the data coming from the input layer. Multiple layers together combined and stacked to make a complete almost perfect model. For example, If an image is given as input then the hidden layer's neuron extract different features from the image like shapes, colors, number of objects and other patters.
- **Output layer** : In this layer the model produce its final output or predictions based on some activation functions. Depending on the activation functions results varies for different problems. For example, to classify binary result like yes or no Sigmoid function is used, for multi-class classification like identifying an animal from multiple animals Softmax is used.

Suppose we have a student dataset where study hour, attendence and exam score are given. Now we want to train a model to predict the exam score based on the study hour and attendance.
Here, at first the numeric values enters into the network. Then input values are multiplied by the weights. After the multiplication each neuron calculates the weighted sum and adds a bias with it. This whole calculation then feeded into an activation function which transforms the result into a predictive value or fixed range and the output layer generates the final prediction. This is how weights, biases, and activation functions work together to transform input data into predictions.
