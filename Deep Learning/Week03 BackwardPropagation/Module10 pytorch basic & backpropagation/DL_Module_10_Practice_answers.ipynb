a) Fromula of $Z_{1}$  :

$Z_{1} = w_{1}x + b_{1}$

b) We know $ReLU$ function as :    
$ReLU(x)= max(0,x)$
for neuron $A_{1}$

$ReLU(A[outputs]) = max(0,A[outputs])$

c) final output formula :     
$\hat{y} = w_{2}A_{1}+b_{2}$

d) From given dimentions  :     
we get,  $x=\begin{bmatrix}
1 \\
2
\end{bmatrix}$

$w_{1} = \begin{bmatrix}
0.1 & 0.1 \\
0.1 & 0.1
\end{bmatrix}$

$w_{2}=\begin{bmatrix}0.1 & 0.1 \end{bmatrix}$

$b_{1}= \begin{bmatrix}
0\\
0
\end{bmatrix}$

$b_{2}=[0]$

now, $Z_{1} = \begin{bmatrix} 0.1 & 0.1 \\ 0.1 & 0.1 \end{bmatrix}
\begin{bmatrix} 1 \\ 2 \end{bmatrix} +
\begin{bmatrix} 0 \\ 0 \end{bmatrix} = \begin{bmatrix} 0.3 \\ 0.3 \end{bmatrix}$
by applying $ReLU$ on $A_{1}$ neuron we get,
 $A_{1} = \begin{bmatrix} 0.3 \\ 0.3 \end{bmatrix}$

 Now for $\hat{y}$ ,

 $\hat{y} = w_{2}A_{1}+b_{2} = \begin{bmatrix}0.1 & 0.1 \end{bmatrix}\begin{bmatrix} 0.3 \\ 0.3 \end{bmatrix} + [0] = 0.06$ .

 Applying $ReLU$ on $\hat{y}$ we get 0.06 as our final output for the network .
