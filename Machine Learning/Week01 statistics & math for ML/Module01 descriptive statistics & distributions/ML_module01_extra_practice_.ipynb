```python
import numpy as np

salaries = np.array([45, 48, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90, 95, 150, 500])

# 1 range
salary_range = np.ptp(salaries)
print(f"Range: {salary_range}")

# varience
varience = np.var(salaries)
print(f"Varience: {varience}")

# standard deviation
standard_deviation = np.std(salaries)
print(f"Standard Deviation: {standard_deviation}")

# Q2
q1 = np.percentile(salaries, 25)
q2 = np.median(salaries)
q3 = np.percentile(salaries, 75)
print(f"Q2: {q2}")

# iqr
iqr = q3 - q1
print(f"IQR: {iqr}")

# lower fence & upper fence
lf = q1 - 1.5 * iqr
uf = q3 + 1.5 * iqr
print(f"Lower Fence: {lf}")
print(f"Upper Fence: {uf}")

# identify outliners
outliners = salaries[(salaries < lf) | (salaries > uf)]
print(f"Outliners: {outliners}")
```
