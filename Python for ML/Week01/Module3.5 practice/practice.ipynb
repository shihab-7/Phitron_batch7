```python
str=input()
if str.find("happy")!=-1 or str.find("joy")!=-1 or str.find("smile")!=-1:
    print("Happy Mood")
elif str.find("sad")!=-1 or str.find("cry")!=-1 or str.find("angry")!=-1:
    print("Sad Mood")
else:
    print("Neutral Mood")
```

```python
list=input().split()
list=[int(i) for i in list]
total=sum(list)
avg=total/len(list)
if avg>170:
    print("Bright Image")
elif avg<85:
    print("Dark Image")
else:
    print("Normal Image")
```

```python
inp = input().split()
cntA = inp.count("A")
cntB = inp.count("B")
if cntA/len(inp)>0.7 or cntB/len(inp)>0.7:
    print("Biased Model")
else:
    print("Fair Model")
```

```python
message = input().split()
lst=["ai","data","model","learn","train","neural"]
cnt=0
for i in lst:
    if i in message:
        cnt+=1
if cnt<2:
    print("Not AI Related")
else:
    print("AI Detected")
```
