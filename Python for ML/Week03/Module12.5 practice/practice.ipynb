```python
students={
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [20, 22, 21, 23],
    'Grade': ['A', 'B', 'A', 'C']
}
df_students = pd.DataFrame(students)

print(df_students)
#drop column Grade
df_students.drop('Grade', axis=1, inplace=True) 
print(df_students)

#drop row with index 1
df_students.drop(1, inplace=True)
print(df_students)
```

```python
employees={
    'Name': ['John', 'Jane', 'Doe', 'Smith'],
    'Department': ['HR', 'Finance', 'IT', 'Marketing'],
    'Salary': [50000, 60000, 55000, 45000]
}
df_employees = pd.DataFrame(employees)
print(df_employees)
sorted = df_employees.sort_values(by=['Salary', 'Name'], ascending=[True, False])
print(sorted)
```

```python
students={
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'score': [5, 40, 38, 92, 61, 55, 82, 91]
}
df_students = pd.DataFrame(students)
# print(df_students)

ans = df_students.loc[(df_students['score'] < 60) & (df_students['score'] > 40),['Name', 'score']]
print(ans)
```
