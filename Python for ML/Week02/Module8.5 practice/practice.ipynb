Smart Vehicle Classifier

```python
class Vehicle:
    def __init__(self, type, name, speed):
        self._type = type #protected
        self.__name = name #private
        self._speed = speed #protected
    
    def classify(self):
        if self._speed >100:
            return "High-speed Vehicle"
        else:
            return "Normal Vehicle"
        
    def show_info(self):
        print(f"Type: {self._type}, Name: {self.__name}, Speed: {self._speed} km/h, Predicted: {self.classify()}")

class Car(Vehicle):
    def __init__(self,name,speed):
        super().__init__("Car",name,speed)
    def show_info(self):
        super().show_info()


class Bike(Vehicle):
    def __init__(self,name,speed):
        super().__init__("Bike",name,speed)
    def show_info(self):
        super().show_info()

n = int(input())
lst=[]
for i in range(n):
    ty,nm,sp= input().split()
    speed=int(sp)
    if ty.lower()=='car':
        lst.append(Car(nm,speed))
        print(f"Car Added: {nm}")
    elif ty.lower()=='bike':
        lst.append(Bike(nm,speed))
        print(f"Bike Added: {nm}")
    else :
        print("Wrong input")

print("\n--- Vehicle Details ---")
for v in lst:
    v.show_info()

    
```
