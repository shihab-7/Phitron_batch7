string

```python
str = "hello world"
print(type(str))
print(str[0:5:2])
# string slicing can be done using the syntax string[start:stop:step]
# in python reverse slicing can be done using negative step
print(str[::-1])
```

methods of string

```python
string = "Machine Learning is fun"

print(len(string))

print("fun" in string)

print(string.upper())
print(string.lower())

# to find index of substring
print(string.find("Learning"))

# find from last
print(string.rfind("in"))

print(string.count("n"))

# to replace substring
new_string = string.replace("fun", "awesome")
print(new_string)
print(string)  # original string remains unchanged because strings are immutable in Python

# google for more string methods in python
```

splitting, joining, formatting

```python
string = "AI is the future"
string2 = "AI-is-the-future"
tokens = string.split()  # split the string into a list of words
print(tokens)

tokens2 = string2.split("-")  # split the string using a custom delimiter
print(tokens2)

sentence = " ".join(tokens)  # join the list of words back into a string
print(sentence)

sentence2 = "-".join(tokens2)  # join the list of words back into a string using a custom delimiter
print(sentence2)
```

formatting string

```python
name = "Alice"
age = 30
account_balance = 1000.5089735262

print(f"My name is {name.upper()}, I am {age} years old and my account balance is ${account_balance:.2f}")
```

list

```python
list1 = [1, 2, 3]
print(type(list1))
# list is mutable, we can change its content
list1[0] = 10
print(list1)

print(list1[0:2])  # slicing a list

list2 = [4, 5, 6]
combined_list = list1 + list2  # concatenating two lists
combined_list.append(100)  # appending an element to the list
print(combined_list)

# insert an element at a specific index
combined_list.insert(1, 50)
print(combined_list)

# remove an element from the list
combined_list.remove(50)
print(combined_list)
# pop an element from the list
combined_list.pop(1)  # pop the element at index 1
print(combined_list)

# multiple data types in a list
mixed_list = [1, "hello", 3.14, [1, 2, 3]]
print(mixed_list)
```

stack and queue with list

```python
lst = [10,1,10,20,15,5]

lst.sort()  # sort the list in ascending order
print(lst)
lst.sort(reverse=True)  # sort the list in descending order
print(lst)

# use list as a stack
stack = []
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)

# lifo
print(stack.pop())  # pop the last element
print(stack.pop())

# use list as a queue
queue = []
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)

# fifo
print(queue.pop(0))  # pop the first element
print(queue.pop(0))
```

list comprehension

```python
even=[]

for i in range(0, 100):
    if i % 2 == 0:
        even.append(i)

print(even)

# list comprehension to generate even numbers from 0 to 99
even_comp = [x for x in range(100) if x % 2 == 0]
# structure of list comprehension is [expression for item in iterable if condition]
print(even_comp)
```

string list comprehension

```python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

capitalized_fruits = [fruit.capitalize() for fruit in fruits]
print(capitalized_fruits)

# without list comprehension
cap_fruits = []
for fruit in fruits:
    cap_fruits.append(fruit.capitalize())
print(cap_fruits)

# everything which are iterable can be used in list comprehension
```
