```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('Practice Day (1).csv')
# df
df["discounted_price"] = (df["Price"]-(df["Price"]*(10/100)))
df.to_csv('discounted_products.csv', index=False)
df
```

```python
df = pd.read_csv('discounted_products.csv')
duplct = df["EmployeeID"].duplicated().sum()
print("Duplicate found: ",duplct)
df.drop_duplicates(inplace=True)
df
```

```python
missing_values = df.isnull().sum()
print(missing_values)
df.fillna({"Age":df["Age"].mean()}, inplace=True)
df
df.drop(df[df["Salary"].isnull()].index, inplace=True)
df
```

```python
mrk1_mean = df["Marks1"].mean()
mrk2_mean = df["Marks2"].mean()
mrk3_mean = df["Marks3"].mean()

print("Marks1 mean: ",mrk1_mean)
print("Marks2 mean: ",mrk2_mean)
print("Marks3 mean: ",mrk3_mean)

mrk1_max = df["Marks1"].max()
mrk2_max = df["Marks2"].max()
mrk3_max = df["Marks3"].max()

print("Marks1 max: ",mrk1_max)
print("Marks2 max: ",mrk2_max)
print("Marks3 max: ",mrk3_max)

df["Total_Marks"] = df["Marks1"] + df["Marks2"] + df["Marks3"]
print(df["Total_Marks"])
```

```python
plt.hist(df["Age"], color='orange', edgecolor='black')
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution Histogram")
plt.grid()
plt.show()

plt.scatter(df["Age"], df["Salary"], color='green', marker='o',edgecolors='black')
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary Scatter Plot")
plt.show()

plt.scatter(df["Price"], df["discounted_price"], color='blue', marker='o',edgecolors='black')
plt.xlabel("Price")
plt.ylabel("Discounted Price")
plt.title("Price vs Discounted Price Scatter Plot")
plt.grid()
plt.show()
```
