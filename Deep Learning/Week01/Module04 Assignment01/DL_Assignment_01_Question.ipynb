# Question 01: [ Marks 10 ]

What is a perceptron?
Explain its three main components and their roles.

### Answer 01:
A perceptron is a simple form of artificial neuron used for binary classification. There are 3 main components , they are :


1. Inputs : These are numeric values of the given data. Each input also carry a weight, which decides the importance of the input predicting the final output.

2. Summation function : This function calculates the weighted sum of all inputs and also adds a bias value , which helps to shift the decision boundary away from the origin.

3. Activation function : When the weighted sum with bias comes here , the activation function coverts it into 0 or 1 based on a threshold.

# Question 04: [ Marks 15 ]

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

### Answer 04:
The weight update rule is : w<sub>new</sub> = w<sub>old</sub> + eta ( y<sub>target</sub> -  y<sub>pred</sub> )* x .
Here , eta is the learning rate .

If the prediction is correct then ( y<sub>target</sub> -  y<sub>pred</sub> ) = 0 , w<sub>new</sub> =  w<sub>old</sub> so no update occurs.

If the prediction is wrong, then the weight is updated in the next iteration because the error = ( y<sub>target</sub> -  y<sub>pred</sub> ) is multiplied with the learning rate and added with the old weight. This is why the weights are only updated for incorrect prediction.

# Question 06: [ Marks 20 ]

What is meant by the “perceptron Learning Rule” or weight adjustment method?
Why can adjusting one weight affect the classification of other points?

### Answer 06:
Perceptron learning rule is an iterative algorithm that adjusts weights based on the error of each training example. It continuously tries to change the weights untill all points are classified correctly.

When we adjust a weight to fix the classification of a point, we shift the entier decision boundary. This shift might move the line to the wrong side of a point which was previously classified correctly. This is why the algorithm requires many epochs to find a balance that satisfies all points.
