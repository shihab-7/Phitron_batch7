```python
vals = input().split()
x= float(vals[0])
min_v= float(vals[1])
max_v= float(vals[2])

norm=(x - min_v) / (max_v - min_v)
print(f"{norm:.2f}")
```

```python
n = int(input())
target=float(input())
sum=0
for i in range(n):
    loss=float(input())
    sum+=loss
avg=sum/n
if avg<=target:
    print("PASS")
else:
    print("RETRY")
```

```python
n = int(input())
y_cnt=0; n_cnt=0

for i in range(n):
    vote=input()
    if vote=="YES":
        y_cnt+=1
    else :
        n_cnt+=1

if y_cnt>=n_cnt:
    print("ACCEPT")
else:
    print("REJECT")
```
